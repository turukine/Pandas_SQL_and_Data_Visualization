```python
import pandas as pd
import sqlite3
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('../data/ab-test.csv')
df.head()
```

```python
df_after = df[(df['time'] == 'after')]
df_after_test = df_after[df_after['group'] == 'test']['diff']
df_after_control =  df_after[df_after['group'] == 'control']['diff']

df_before = df[(df['time'] == 'before')]
df_before_test = df_before[df_before['group'] == 'test']['diff']
df_before_control = df_before[df_before['group'] == 'control']['diff']
```

```python
labels = ['control','test']
plt.figure(figsize=(15,8))
plt.suptitle('Lag before labs deadlines. After and before the Newsfeed', fontsize=15)
plt.axis('off')

plt.subplot(1, 2, 1)
plt.boxplot(
    [df_after_control, df_after_test],
    tick_labels=labels,
    boxprops=dict(color='green', linewidth=3),
    whiskerprops=dict(color='red'),
    capprops=dict(color='red'),
    medianprops=dict(color='red', linewidth=2),
    showmeans=True
           );
plt.grid(linestyle='--', alpha=0.7)
plt.title('after', fontsize=15)

plt.subplot(1, 2, 2)
plt.boxplot(
    [df_before_control, df_before_test],
    boxprops=dict(color='green', linewidth=3),
    tick_labels=labels,
    whiskerprops=dict(color='red'),
    capprops=dict(color='red'),
    medianprops=dict(color='red', linewidth=2),
    showmeans=True
           );
plt.grid(linestyle='--', alpha=0.7)
plt.title('before', fontsize=15)
plt.show()
```
